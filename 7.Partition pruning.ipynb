```python
from pyspark.sql import SparkSession


# Step : Sample data
data = [
    ("Alice", "HR", 1000),
    ("Bob", "IT", 2000),
    ("Charlie", "HR", 1500),
    ("David", "Finance", 2500),
    ("Eve", "IT", 1800),
    ("Frank", "Finance", 2200)
]

columns = ["name", "department", "salary"]

# Step : Create DataFrame
df = spark.createDataFrame(data, columns)

# Step : Write DataFrame using partitioning
output_path = "/Volumes/workspace/default/my_files/OutputData"  

df.write \
  .mode("overwrite") \
  .partitionBy("department") \
  .parquet(output_path)

```

```python
output_path_new = "/Volumes/workspace/default/my_files/OutputDataWithout"  

df.write \
  .mode("overwrite") \
  .parquet(output_path_new)
```

```python
from pyspark.sql.functions import *
```

```python
df_withpart = spark.read.format("parquet")\
                .load(output_path)
```

```python
display(df_withpart)
```

```python
df_withoutpart = spark.read.format("parquet")\
                .load(output_path_new)\
                .filter(col('department')=='HR')
```

```python
display(df_withoutpart)
```

**Dynamic Partition Pruning**

```python
df_join_new = df_withoutpart.join(df_withpart, (df_withoutpart['name']==df_withpart['name']) & (df_withoutpart['department']==df_withpart['department']), how='inner')
```
